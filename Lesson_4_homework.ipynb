```python
#Exercise 9
a = 75
b = 50
c = 100
d = 62.5
min_value = b
max_value = c
a_normalized = (a - min_value) / (max_value - min_value)
b_normalized = (b - min_value) / (max_value - min_value)
c_normalized = (c - min_value) / (max_value - min_value)
d_normalized = (d - min_value) / (max_value - min_value)
column_heading_a = 'Original Value'
column_heading_b = 'Normalized Value'
print(f'{column_heading_a:^20}{column_heading_b:^20}')
print(f'{a:^20}{a_normalized:^20}')
print(f'{b:^20}{b_normalized:^20}')
print(f'{c:^20}{c_normalized:^20}')
print(f'{d:^20}{d_normalized:^20}')


```
