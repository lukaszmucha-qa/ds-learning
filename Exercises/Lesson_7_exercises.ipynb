```python
#Exercise 9
weight_data = [55, 60, 70, 85, 62, 95, 78]
height_data = [1.75, 1.70, 1.75, 1.80, 1.65, 1.78, 1.72]
bmi = []
for iteration in range(len(weight_data)):
    bmi.append(round(weight_data[iteration] / (height_data[iteration] ** 2), 2))
    if bmi[iteration] >= 25:
        print(f"Customer with weight {weight_data[iteration]} kg and height {height_data[iteration]} m has BMI of {bmi[iteration]} and is overweight.")
    elif bmi[iteration] < 18.5:
        print(f"Customer with weight {weight_data[iteration]} kg and height {height_data[iteration]} m has BMI of {bmi[iteration]} and is underweight.")
    else:
        print(f"Customer with weight {weight_data[iteration]} kg and height {height_data[iteration]} m has BMI of {bmi[iteration]} and is in normal range.")

```

```python
#Exercise 10
list_of_prime_numbers = []
for n in range(2, 101):
    prime = True
    for m in range(2, int((n ** 0.5 + 1))):
        if n % m == 0:
            prime = False
    if prime:
        list_of_prime_numbers.append(n)
print(list_of_prime_numbers)
```
