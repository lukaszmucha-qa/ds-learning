```python
#Exercise 5
string_to_process = "Alice,28,Data Scientist,75000"
processed_string = str(string_to_process).split(",")
print(processed_string)
processed_string_v_2 = []
for element in processed_string:
    if element.isdigit():
        processed_string_v_2.append(int(element))
    else:
        processed_string_v_2.append(element)
print(processed_string_v_2)
print(f'{processed_string[0]} ma {processed_string[1]} lat, pracuje jako {processed_string[2]} i zarabia ${processed_string[3]}')
```

```python
#Exercise 8
file_name = 'model_results_2024_Q3.csv'
index_of_digit = []
for char in file_name:
    if char.isdigit():
        index_of_digit.append(file_name.index(char))
year_and_quarter = file_name[index_of_digit[0]:index_of_digit[4]+1]
year_and_quarter = year_and_quarter.split("_")
print(f'Year: {year_and_quarter[0]}, Quarter: {year_and_quarter[1]}')
```
