```python
#Exercise 9
original_dictionary = {'laptop' : 'electronics' , 'keyboard' : 'electronics' , 'desk' : 'furniture' , 'chair' : 'furniture' }
list_of_values = original_dictionary.values()
dictionary_keys = list(set(list_of_values))
reversed_dictionary = dict.fromkeys(dictionary_keys)

print(dictionary_keys)
print(reversed_dictionary)

for x in reversed_dictionary:
   temp_list = []
   for key, value in original_dictionary.items():
       if x == value:
           temp_list.append(key)
   reversed_dictionary[x] = temp_list
print(reversed_dictionary)
```

```python
#Exercise 11
my_dict = {str(x): x**2  for x in range(1, 21)}
print(f'Dictionary with values to the power of 2: {my_dict}')
my_dict = {key: value for key, value in my_dict.items() if int(key) % 2 == 0}
print(f'Dictionary with only even keys: {my_dict}')
```
